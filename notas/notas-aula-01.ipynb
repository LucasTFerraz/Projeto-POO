## 3. Booleanos, if/elif/else e a armadilha dos ifs independentes

Comparações (`>`, `<`, `==`, `!=`) devolvem `True`/`False`. `and`/`or`/`not` são palavras em Python, não símbolos (`&&`/`||`/`!`). Python também permite encadear comparações como na matemática: `0 <= x < 10`.

`if`/`elif`/`else` formam uma escada **mutuamente exclusiva** — para no primeiro verdadeiro. Vários `if` soltos são **independentes** — todos rodam. No robô, a checagem `0 <= novo_x < LADO_GRADE` é a nossa condição de "dentro da grade".


```python
# Comparações e operadores lógicos
print(5 > 3)                  # True
print(10 == 10.0)              # True — int e float iguais em valor
print(0 <= 7 < LADO_GRADE)     # True — encadeamento Python, dentro da grade

# and/or/not são PALAVRAS em Python, não &&/||/!
print((5 > 3) and (2 > 10))    # False
print((5 > 3) or (2 > 10))     # True
```

```python
# A armadilha: ifs independentes (todos rodam) vs. escada if/elif/else (para no primeiro)
nota = 95

print("--- ifs independentes ---")
if nota >= 50:
    print("Aprovado")
if nota >= 70:
    print("Com mérito")
if nota >= 90:
    print("Com louvor")     # as TRÊS linhas rodam para nota=95!

print("--- escada if/elif/else ---")
if nota >= 90:
    print("Com louvor")
elif nota >= 70:
    print("Com mérito")
elif nota >= 50:
    print("Aprovado")
else:
    print("Reprovado")       # só UMA linha roda
```

## 4. Robô v1 — decide antes de andar (dois eixos)

Antes de mover, o robô calcula o **destino** (`novo_x`) e só então decide se aceita — nunca atualiza a posição antes de checar. A mesma checagem se repete para os dois eixos.


```python
# Robô v1: decide no eixo x
x, y = 0, 0
passos = 8
novo_x = x + passos          # calcula o destino ANTES de decidir
if novo_x < 0:
    print("Parede oeste!")
elif novo_x >= LADO_GRADE:
    print("Parede leste!")
else:
    x = novo_x
    print(f"Robô em ({x}, {y})")
```

```python
# Robô v1: decide nos dois eixos (x e y)
x, y = 0, 0

passos_x = 8
novo_x = x + passos_x
if novo_x < 0:
    print("Parede oeste!")
elif novo_x >= LADO_GRADE:
    print("Parede leste!")
else:
    x = novo_x

passos_y = 3
novo_y = y + passos_y
if novo_y < 0:
    print("Parede sul!")
elif novo_y >= LADO_GRADE:
    print("Parede norte!")
else:
    y = novo_y

print(f"Robô em ({x}, {y})")
```

## 5. `for`: range, listas e padrões de leitura de lote

`range(n)` gera de `0` a `n-1` — o limite é **exclusivo**. `for item in lista` (sem índice) é mais Pythônico que `for i in range(len(lista))` quando o índice não importa.

No robô, o padrão "iterando lista" é o que o v2 vai usar para executar uma sequência de comandos.


```python
# range: fim é exclusivo
for i in range(5):
    print(i)              # 0 1 2 3 4

# for em lista — sem índice, mais Pythônico
comandos = ["AVANCAR 3", "GIRAR ESQ", "AVANCAR 2", "PARAR"]
for cmd in comandos:
    if cmd == "PARAR":
        break
    print(f"  executando: {cmd}")
```

## 6. `while`: repetição indefinida

`for` serve quando sabemos quantas iterações; `while` quando não sabemos — a condição de parada depende do que acontece durante o laço. **Cuidado:** `while True` sem um `break` em algum ramo é um **loop infinito** — em aula, isso é demonstrado ao vivo e interrompido com `Ctrl+C`; aqui, sempre usamos `while True` **com** uma condição de saída clara.

No robô v2, o padrão sentinela (`while True` + `break` ao encontrar `"PARAR"`) é uma forma de simular a leitura de comandos até o sinal de parada.


```python
# Padrão sentinela: processar uma fila de comandos até 'PARAR'
fila = ["AVANCAR 3", "GIRAR ESQ", "PARAR", "AVANCAR 99"]
i = 0
while True:
    cmd = fila[i]
    if cmd == "PARAR":
        print("Programa encerrado.")
        break
    print(f"  executando: {cmd}")
    i += 1
```

## 7. Strings e parsing

`split()` divide uma string em pedaços por espaço; `strip()` remove espaços nas bordas; `upper()`/`lower()` normalizam maiúsculas/minúsculas. O padrão do parser do robô é `strip().upper().split()` — normaliza tudo antes de decidir.

No robô, `"AVANCAR 3"` vira `["AVANCAR", "3"]`; o segundo item ainda é **string** — precisa de `int(...)` para virar número.


```python
# Normalizar e dividir
comando = "  avancar   3  "
partes = comando.strip().upper().split()
print(partes)                  # ['AVANCAR', '3']

acao = partes[0]
valor = int(partes[1])         # split devolve str — converter com int!
print(acao, valor)

print("AVAN" in "AVANCAR")   # True — checar substring
```

## 8. Robô v2 — executa uma sequência de comandos

O v2 junta tudo: lê uma lista de comandos, faz o parser normalizar cada um, e despacha por ação (`AVANCAR`/`GIRAR`/`PARAR`), reaproveitando a checagem de parede do v1 para os dois eixos.


```python
LADO_GRADE = 10
x, y = 0, 0
direcao = "LESTE"

comandos = ["AVANCAR 3", "GIRAR ESQ", "AVANCAR 2", "GIRAR DIR", "AVANCAR 5", "PARAR"]

for cmd in comandos:
    partes = cmd.strip().upper().split()
    acao = partes[0]
    if acao == "PARAR":
        print("Robô parou.")
        break
    elif acao == "AVANCAR":
        passos = int(partes[1])
        if direcao == "LESTE":
            novo_x = x + passos
            if 0 <= novo_x < LADO_GRADE:
                x = novo_x
            else:
                print("  Parede! Ficou em x =", x)
        elif direcao == "NORTE":
            novo_y = y + passos
            if 0 <= novo_y < LADO_GRADE:
                y = novo_y
            else:
                print("  Parede! Ficou em y =", y)
        elif direcao == "OESTE":
            novo_x = x - passos
            if 0 <= novo_x < LADO_GRADE:
                x = novo_x
            else:
                print("  Parede! Ficou em x =", x)
        elif direcao == "SUL":
            novo_y = y - passos
            if 0 <= novo_y < LADO_GRADE:
                y = novo_y
            else:
                print("  Parede! Ficou em y =", y)
    elif acao == "GIRAR":
        lado = partes[1]
        if lado == "ESQ":
            if direcao == "LESTE":    direcao = "NORTE"
            elif direcao == "NORTE":  direcao = "OESTE"
            elif direcao == "OESTE":  direcao = "SUL"
            elif direcao == "SUL":    direcao = "LESTE"
        elif lado == "DIR":
            if direcao == "LESTE":    direcao = "SUL"
            elif direcao == "SUL":    direcao = "OESTE"
            elif direcao == "OESTE":  direcao = "NORTE"
            elif direcao == "NORTE":  direcao = "LESTE"
    print(f"  [{acao}] Robô em ({x}, {y}), direção {direcao}")
```
