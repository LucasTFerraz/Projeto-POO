## `dict` vs. `class` — de onde vem o `Robo`

Até a Aula 4 (D1), o robô era um `dict`: `robo["x"]`, `robo["direcao"]`. Só que `dict` é um
tipo genérico do Python — ele nunca soube que aquilo representava um robô. `class` cria um
**tipo novo**: dentro dela, `__init__` roda automaticamente ao criar um objeto, guardando
cada atributo com `self.atributo = valor`. Depois de instanciado, acessamos por ponto
(`robo1.x`), não por colchete.

```python
robo_dict = {"x": 0, "y": 0, "direcao": "LESTE"}
print(robo_dict)
print(type(robo_dict))
```

```python
class Robo:
    def __init__(self, x=0, y=0, direcao="LESTE"):
        self.x = x
        self.y = y
        self.direcao = direcao

robo1 = Robo()
print(type(robo1))
print(robo_dict["x"], robo1.x)
```

## `self` — o parâmetro que o Python passa sozinho

Quando vocês escrevem `objeto.metodo(...)`, o Python reescreve por trás dos panos para
`Classe.metodo(objeto, ...)` — o objeto vira o primeiro argumento, automaticamente. Por isso
todo método precisa de um primeiro parâmetro para recebê-lo; por convenção universal, ele se
chama `self` (o nome não importa, só a posição). No robô, é assim que `avancar(robo, ...)` da
D1 virou `robo.avancar(...)` na D2: `self` é o `robo` de sempre, só que a linguagem passa ele
para vocês.

```python
class Bateria:
    def __init__(self, nivel=100):
        self.nivel = nivel

    def relatorio(self):
        return f"Bateria em {self.nivel}%"

b1 = Bateria(80)
print(b1.relatorio())
print(Bateria.relatorio(b1))
```

```python
class Exemplo:
    def falar(coisa):             # nomeei de "coisa", não "self" — funciona igual
        print("funcionou! o primeiro parâmetro não precisa se chamar 'self'")

Exemplo().falar()


class SemSelf:
    def cumprimentar():          # esqueceu o self
        print("oi")

try:
    SemSelf().cumprimentar()
except TypeError as erro:
    print(f"TypeError: {erro}")
```

## Atributos de classe vs. atributo de instância

Um atributo pode morar em dois lugares: dentro do `__init__` (`self.atributo = ...`) — aí é
de **instância**, uma cópia por objeto; ou direto no corpo da classe, fora de qualquer
método — aí é de **classe**, uma cópia só, compartilhada por todos os objetos. No robô,
`LADO_GRADE` (o tamanho da grade) é de classe — não muda de robô para robô; `x`/`y`/`direcao`
são de instância — cada robô tem os seus.

```python
class Robo:
    LADO_GRADE = 5          # atributo de classe

    def __init__(self, x=0, y=0):
        self.x = x           # atributo de instância
        self.y = y

r1 = Robo()
r2 = Robo(3, 3)
print(r1.LADO_GRADE, r2.LADO_GRADE)
print(r1.x, r2.x)

Robo.LADO_GRADE = 8
print(r1.LADO_GRADE, r2.LADO_GRADE)
```

## Cuidado com valor default mutável

Um valor default de parâmetro é criado **uma única vez**, quando o Python lê a definição do
método — não uma vez por chamada. Se esse valor for uma lista ou dict, todas as instâncias
criadas sem passar aquele argumento vão compartilhar o **mesmo** objeto. No robô, isso mordeu
quando `obstaculos={}` era o default do `__init__`: dois robôs sem obstáculos próprios
acabavam enxergando o mesmo dict. O conserto: usar `None` como sentinela e criar a coleção
nova dentro do `__init__`.

```python
class ComBug:
    def __init__(self, itens=[]):
        self.itens = itens

a = ComBug()
b = ComBug()
a.itens.append("x")
print(b.itens)
print(id(a.itens) == id(b.itens))


class Corrigido:
    def __init__(self, itens=None):
        self.itens = itens if itens is not None else []

c = Corrigido()
d = Corrigido()
c.itens.append("x")
print(d.itens)
```

## `avancar`/`girar`: comportamento dentro do objeto

Na D1, `avancar(robo, obstaculos)` e `girar(robo, lado)` eram funções soltas — cada uma
recebia `robo` para saber sobre qual objeto agir. Como método, o mesmo comportamento fica
**dentro** da classe, e `self` substitui o `robo` que vocês já passavam à mão: `self.x`,
`self.direcao`, sem precisar receber nada disso como argumento.

```python
class Robo:
    DELTAS = {"LESTE": (1, 0), "NORTE": (0, 1), "OESTE": (-1, 0), "SUL": (0, -1)}
    GIRAR_ESQ = {"LESTE": "NORTE", "NORTE": "OESTE", "OESTE": "SUL", "SUL": "LESTE"}
    GIRAR_DIR = {"LESTE": "SUL", "SUL": "OESTE", "OESTE": "NORTE", "NORTE": "LESTE"}

    def __init__(self, x=0, y=0, direcao="LESTE"):
        self.x = x
        self.y = y
        self.direcao = direcao

    def avancar(self):
        dx, dy = Robo.DELTAS[self.direcao]
        self.x += dx
        self.y += dy

    def girar(self, lado):
        if lado == "ESQ":
            self.direcao = Robo.GIRAR_ESQ[self.direcao]
        elif lado == "DIR":
            self.direcao = Robo.GIRAR_DIR[self.direcao]


robo1 = Robo()
robo1.girar("ESQ")
robo1.avancar()
robo1.avancar()
print(robo1.x, robo1.y, robo1.direcao)
```

## Vários robôs, o mesmo comportamento, estados diferentes

Cada `Robo(...)` que vocês instanciam é independente — `self.x` de um robô nunca é o `self.x`
de outro. O método é o mesmo (definido uma vez na classe); o estado (`x`, `y`, `direcao`) é de
cada objeto.

```python
class Robo:
    DELTAS = {"LESTE": (1, 0), "NORTE": (0, 1), "OESTE": (-1, 0), "SUL": (0, -1)}
    GIRAR_ESQ = {"LESTE": "NORTE", "NORTE": "OESTE", "OESTE": "SUL", "SUL": "LESTE"}
    GIRAR_DIR = {"LESTE": "SUL", "SUL": "OESTE", "OESTE": "NORTE", "NORTE": "LESTE"}

    def __init__(self, x=0, y=0, direcao="LESTE"):
        self.x = x
        self.y = y
        self.direcao = direcao

    def avancar(self):
        dx, dy = Robo.DELTAS[self.direcao]
        self.x += dx
        self.y += dy

    def girar(self, lado):
        if lado == "ESQ":
            self.direcao = Robo.GIRAR_ESQ[self.direcao]
        elif lado == "DIR":
            self.direcao = Robo.GIRAR_DIR[self.direcao]


robo_a = Robo(x=0, y=0, direcao="LESTE")
robo_b = Robo(x=9, y=9, direcao="OESTE")

robo_a.avancar()
robo_b.girar("ESQ")
robo_b.avancar()

print(robo_a.x, robo_a.y, robo_a.direcao)
print(robo_b.x, robo_b.y, robo_b.direcao)
```
